```python
import pandas as pd
import numpy as np
```

```python
df_train = pd.read_csv("train.csv")
df_valid = pd.read_csv("valid.csv")
df_test = pd.read_csv("test.csv")

df_full = pd.concat([df_train, df_valid, df_test])
df_full.head()
```

```python
df_full.iloc[0]["filename"]
```

```python
df_full = df_full.filter(items=["filename", "class", "xmin", "ymin", "xmax", "ymax"])
df_full.columns = ["image_filename", "class", "x1", "y1", "x2", "y2"]
df_full.head()
```

```python
current_fname = df_full.iloc[0]["image_filename"][:-28]
current_fname = current_fname[:-4] + ".png"
current_fname
```

```python
filenames_formatted = []
for i in range(len(df_full)):
    current_fname = df_full.iloc[i]["image_filename"][:-28]
    current_fname = current_fname[:-4] + ".png"
    filenames_formatted.append(current_fname)

df_full["image_filename"] = filenames_formatted
df_full.head()
```

```python
uniq_starts = set([f[:-8] for f in filenames_formatted])
uniq_starts
```

```python
df_formatted = df_full
contains_start = ['glw_ac_25c_3_dpf_fert_04_24_2026' in f for f in filenames_formatted]
df_formatted["contains_start"] = contains_start
df_formatted.shape
```

```python
df_formatted[df_formatted["contains_start"] == True]
```

```python
for start in uniq_starts:
    df_formatted = df_full
    contains_start = [start in f for f in filenames_formatted]
    df_formatted["contains_start"] = contains_start
    df_formatted = df_formatted[df_formatted["contains_start"] == True]
    df_formatted = df_formatted.filter(items=["image_filename", "class", "x1", "y1", "x2", "y2"])

    print("Saving " + start)
    df_formatted.to_csv(start + "_annotations.csv", index=False)
```
